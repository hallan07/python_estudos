```python
# Aprendendo classes

class Carro:
  """ __init__ inicia automaticamente ao chamar a classe, nesse caso, cria atributos """
  def __init__(self, marca, cor, cc):
    self.marca = marca
    self.cor = cor
    self.cc = cc
  
  """ Aqui as funções/métodos fazem uma ação """
  def ligar(self):
    print(f"O {self.marca} está ligado!")
    
  def desligar(self):
    print(f"O {self.marca} está desligado...")
    
  def buzinar(self):
    print("Biiiii bib!")
    
  def espec(self):
    print(f"marca: {self.marca}\ncor: {self.cor}\ncc: {self.cc}")
```

```python
# Executando a classe

# Crando objetos
meu_carro = Carro("Toyota", "vermelho", "1.5")
teu_carro = Carro("Ford", "azul", "1.3")
```
